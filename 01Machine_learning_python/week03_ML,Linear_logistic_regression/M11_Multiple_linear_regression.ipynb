```python
def make_pred(X,y,W,b):
    # calculate f(x) = wx + b
    m=X.shape[0]

    pred_list = np.zeros((m,))

    for i in range(m):
        # X[i] = feature vector, W= weights vector
        pred_list[i] = np.dot(W, X[i]) + b

    return pred_list
```

```python
m = X.shape[0]
n = X.shape[1]
n
```

```python
W_init = np.ones((n,))
b_init = 1.0

make_pred(X, y, W_init, b_init)
```

```python
def compute_cost(X,y,W,b):
    m=X.shape[0] # no of data points

    cost =0.0
    pred_list = make_pred(X,y,W,b)

    print(pred_list)

    error = pred_list - y
    print(error)
    error_squared = error ** 2
    print(error_squared)
    cost = np.sum(error_squared)

    cost = cost / (2*m)

    return cost
```

```python
compute_cost(X,y, W_init, b_init)
```
