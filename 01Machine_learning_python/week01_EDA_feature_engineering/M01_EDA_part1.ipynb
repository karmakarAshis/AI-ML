```python
sns.histplot(data=df, x='Age', bins=50)
plt.title('Histogram of Age')
plt.xlabel('Ages')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.kdeplot(data=df, x='Age')
plt.title('KDE Plot of Age')
plt.xlabel('Ages')
plt.ylabel('Probability Distribustion')
plt.show()
```

```python
sns.histplot(data=df, x='Fare', bins=50)
plt.title('Histogram of Fare')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.kdeplot(data=df, x='Fare')
plt.title('KDE Plot of Fare')
plt.xlabel('Fare')
plt.ylabel('Probability Distribustion')
plt.show()
```
